# BTVN 03;
</br> Tạo một tập dữ liệu gia lập vói x la số giờ học(ngẫu nhiên từ 1 - 10 ) va y la số điểm dược tính thep. Voi noise la mot gia tri nga nhien nho
</br> 1 Khoi tao tham so x va b ngau nhien voi requires=true 2 
</br> 2 Tinh MSE
</br> 3 Tinh Gradient 
</br> 4 Tinh cap nhat tham so w va b = gradient
</br> 5 Lap lai qua trinh tren trong 101 lan de xem su hoi tu cua mo hinh

```python
#BTVN 2

import torch
#Khởi tạo x 
x = torch.tensor(2.0, requires_grad=True)
#Learning rate
alpha = 0.1
print("Giá trị x ban đầu:", x.item())
print("-" * 40)
#Gradient Descent trong 10 vòng lặp
for i in range(10):
    #Định nghĩa hàm y
    y = x**3 + 2*x**2 + 5*x + 1
    #Tính gradient
    y.backward()
    with torch.no_grad():
        x -= alpha * x.grad
    x.grad.zero_()
    #In kết quả mỗi vòng
    print(f"Vòng {i+1}: x = {x.item():.6f}, y = {y.item():.6f}")
```

```python
#BTVN 3
import torch
import random

torch.manual_seed(42)

x = torch.rand(50, 1) * 9 + 1       
noise = torch.randn(50, 1) * 0.5 
y = 2 * x + 3 + noise             

w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

#Learning rate
alpha = 0.01

for epoch in range(101):
    #Dự đoán
    y_pred = w * x + b
    
    #Tính MSE
    loss = torch.mean((y_pred - y) ** 2)
    
    #Tính gradient
    loss.backward()
    
    #Cập nhật tham số
    with torch.no_grad():
        w -= alpha * w.grad
        b -= alpha * b.grad
    
    #Reset gradient
    w.grad.zero_()
    b.grad.zero_()
    
    #In kết quả mỗi 10 vòng
    if epoch % 10 == 0:
        print(f"Epoch {epoch:3d}: Loss = {loss.item():.4f}, w = {w.item():.4f}, b = {b.item():.4f}")
```

### BTVN 05
Bạn giúp thầy tạo tensor với 
</br> empyty

</br> ones

```python
import torch

#Tạo tensor với empty
x_empty = torch.empty(3, 4)

print(x_empty)
print(x_empty.shape)

#Tạo tensor với zeros
x_zeros = torch.zeros(3, 4)

print(x_zeros)
print(x_zeros.shape)

#Tạo tensor với ones
x_ones = torch.ones(3, 4)

print(x_ones)
print(x_ones.shape)

#Tạo tensor với random
x_rand = torch.rand(3, 4)

print(x_rand)
print(x_rand.shape)

#Reshape với view
x = torch.arange(0, 12)

print(x)
print(x.shape)

x_view = x.view(3, 4)
print(x_view)
print(x_view.shape)

#Reshape với view_as
y = torch.zeros(3, 4)

x_view_as = x.view_as(y)

print(x_view_as)
print(x_view_as.shape)
```
