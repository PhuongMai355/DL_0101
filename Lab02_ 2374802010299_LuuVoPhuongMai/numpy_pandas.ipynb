```python
# BTVN thú vị

# Cho ma trận = [[99, 99, 99],
#            [99, 99, 99],
#            [99, 99, 99]]

# Hãy tưởng tượng ô 99 trống
# Giả sử: 1 là x
# Giả sử: 0 là o

# Lấy đầu vào từ phía x và phía o luân phiên
# Nếu đầu vào từ phía x alf o,o

# Thì ma trận sẽ trở thành 
# [[x, 99, 99],
# [99, 99, 99],
# [99, 99, 99]]

# Nếu đầu vào từ phía o cũng là (o, o)
# Thông báo họ nhập lại
# Nếu không thì điền vào ma trận

# Thử thách: Nếu người đầu tiên có 3 phần tử liên tiếp hãy dừng trò chơi

# BTVN thú vị: X/O trên ma trận 3x3
# 99 là ô trống
# x = 1
# o = 0
# nhập (row,col) luân phiên
# nếu ô đã có người đánh => nhập lại
# nếu ai có 3 phần tử liên tiếp => dừng game

import numpy as np

board = np.array([
    [99, 99, 99],
    [99, 99, 99],
    [99, 99, 99]
])

current = 1   # x đi trước (1), o là 0

print("Game bắt đầu! Nhập dạng row,col (0..2). Ví dụ: 0,0")

while True:
    # in board: 99 -> .
    show = board.copy().astype(object)
    show[show == 99] = "."
    show[show == 1] = "x"
    show[show == 0] = "o"
    print(show)

    player_char = "x" if current == 1 else "o"
    s = input(f"Lượt {player_char}, nhập (row,col): ").strip()

    # tách row,col
    try:
        r, c = s.split(",")
        r = int(r.strip())
        c = int(c.strip())
    except:
        print("Sai định dạng! Nhập lại kiểu row,col (ví dụ 1,2)")
        continue

    # kiểm tra biên
    if r < 0 or r > 2 or c < 0 or c > 2:
        print("Tọa độ không hợp lệ! Chỉ 0..2. Nhập lại!")
        continue

    # kiểm tra ô trống
    if board[r, c] != 99:
        print("Ô này đã có người đánh rồi. Nhập lại!")
        continue

    # điền vào ma trận
    board[r, c] = current

    # kiểm tra thắng (3 liên tiếp)
    win = False

    # hàng / cột
    for i in range(3):
        if np.all(board[i, :] == current):
            win = True
        if np.all(board[:, i] == current):
            win = True

    # chéo chính / chéo phụ
    if np.all(np.diag(board) == current):
        win = True
    if np.all(np.diag(np.fliplr(board)) == current):
        win = True

    if win:
        show = board.copy().astype(object)
        show[show == 99] = "."
        show[show == 1] = "x"
        show[show == 0] = "o"
        print(show)
        print(f"🎉 {player_char.upper()} thắng! Dừng trò chơi.")
        break

    # kiểm tra hòa (hết ô trống)
    if not np.any(board == 99):
        show = board.copy().astype(object)
        show[show == 99] = "."
        show[show == 1] = "x"
        show[show == 0] = "o"
        print(show)
        print("🤝 Hòa! Bảng đã đầy.")
        break

    # đổi lượt
    current = 0 if current == 1 else 1

```
